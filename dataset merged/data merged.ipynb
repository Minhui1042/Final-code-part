```python
from pathlib import Path

OUTPUT_DIR = Path(
    r"C:\Users\15195\Desktop"
)

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
```

```python
# Create transaction-product merged dataset
transaction_product = transaction.merge(
    product,
    on="PRODUCT_ID",
    how="left",
    validate="many_to_one"
)

print("transaction_product shape:", transaction_product.shape)
print("Missing product classifications after merge:")

print(
    transaction_product[
        ["DEPARTMENT", "COMMODITY_DESC", "SUB_COMMODITY_DESC"]
    ].isnull().sum()
)
```

```python
# Save transaction-product merged dataset
transaction_product.to_csv(
    OUTPUT_DIR / "transaction_product_merged.csv",
    index=False
)

print("Saved:", OUTPUT_DIR / "transaction_product_merged.csv")
```

```python
# Create transaction-demographic merged dataset
transaction_demographic = transaction.merge(
    hh_demographic,
    on="household_key",
    how="left",
    validate="many_to_one",
    indicator=True
)

print("transaction_demographic shape:", transaction_demographic.shape)

print("\nMerge result by transaction row:")
print(transaction_demographic["_merge"].value_counts())
```

```python
#  Check demographic coverage at household level
household_demo_coverage = (
    transaction_demographic[
        ["household_key", "_merge"]
    ]
    .drop_duplicates()
    .groupby("_merge", observed=False)["household_key"]
    .nunique()
)

print("\nMerge result by unique household:")
print(household_demo_coverage)
```

```python
# Save the full transaction-level merge only if you need it for checking
transaction_demographic.to_csv(
    OUTPUT_DIR / "transaction_demographic_merged.csv",
    index=False
)

print("Saved:", OUTPUT_DIR / "transaction_demographic_merged.csv")
```
