```python
import pandas as pd
import re
```

```python
# Reading dataset
rules= pd.read_csv("C:/Users/15195/Desktop/coding part/Apriori part/category_rule_summary.csv")
```

```python
# Assign an ID to each unique association rule (without considering the order of occurrence)
rules['rule_id'] = rules.apply(
    lambda row: ']'.join(sorted([row['antecedents'], row['consequents']])),
    axis=1
)
```

```python
# Group the clusters and count the set of rule IDs for each cluster
cluster_rules = {}
for cluster in rules['cluster'].unique():
    cluster_rules[cluster] = set(rules[rules['cluster'] == cluster]['rule_id'])
```

```python
# Count the number of unique rules (rules present in this cluster but not in other clusters)
unique_counts = {}
for cluster, rule_set in cluster_rules.items() :
    # Merge all the rules from other clusters
    other_rules = set()
    for other_cluster, other_set in cluster_rules.items():
        if other_cluster != cluster:
           other_rules = other_rules.union(other_set)
    # Difference set = unique rules specific to this cluster
    unique_rules = rule_set - other_rules
    unique_counts[cluster] = len(unique_rules)
```

```python
# Print the result
print("Unique Rules by Segment:")

for cluster, count in unique_counts.items():
    print(f"Cluster {cluster}: {count} unique rules")
```
